```python
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs

# Krok 1: Wygenerowanie 3 wyraźnych chmur punktów w przestrzeni 2D
# centers=3 oznacza, że w danych istnieją 3 naturalne skupiska
X, _ = make_blobs(n_samples=300, centers=3, cluster_std=0.8, random_state=42)

# Krok 2: Inicjalizacja i dopasowanie modelu
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
kmeans.fit(X)

# Krok 3: Pobranie wyników
labels = kmeans.labels_  # przypisanie każdego punktu do grupy (0, 1 lub 2)
centroids = kmeans.cluster_centers_  # współrzędne znalezionych środków

# Krok 4: Wizualizacja
plt.figure(figsize=(8, 5))
plt.scatter(
    X[:, 0], X[:, 1], c=labels, cmap="viridis", alpha=0.6, label="Punkty"
)
plt.scatter(
    centroids[:, 0],
    centroids[:, 1],
    c="red",
    s=200,
    marker="X",
    label="Centroidy",
)
plt.title("Wynik klastrowania algorytmem k-means")
plt.xlabel("Cecha 1")
plt.ylabel("Cecha 2")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.5)
plt.show()
```

```python
from sklearn.metrics import silhouette_score

score = silhouette_score(X, kmeans.labels_)
print(f"Współczynnik Silhouette: {score:.3f}")
```
